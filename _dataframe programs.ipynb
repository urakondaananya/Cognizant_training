```python
import pandas as pd
data=pd.DataFrame({
      'Name':['saru','Anu','bunny'],
       'ID':[20,45,19]})
    
data2=pd.DataFrame({
    'marks':[87,94,86,45],
     'ID':[20,87,67,68]})

result=pd.merge(data,data2,on="ID",how="left")
print(result)
```

```python
pd.merge(data,data2,how="inner",on="ID")
```
