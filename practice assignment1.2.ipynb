```python
class Student:      #inheritance
    def __init__(self,name,age):
        self.name=name
        self.age=age

s1=Student("ananya",19)
print(s1.name)
print(s1.age)
```

```python
class Employee:
    def __init__(self,name,id,salary):
        self.name=name
        self.salary=salary
        self.id=id

s1=Employee("ananya",9,80000)
print(s1.name)
print(s1.salary)
print(s1.id)
```

```python
class GrandFather:
    def func(self):
        print("this is GrandFather")

class Father(GrandFather):
    def func(self):
        super().func()
        print("this is father")

class Child(Father):
    def func(self):
        super().func()
        print("this is child")

object=Child()
object.func()
```

```python
def factorial(i):   #functions
    fact=1
    if i==0:
        return 1
    else:
        for j in range(1,i+1):
            fact=fact*j
        return fact

n=4
factorial(n)
```

```python
def prime(n):
    
    if n % 2 ==0:
        print('not prime')
    else:
        print('prime')

j=6
prime(j)
```

```python
def vowels(name):
    
    vowels="aeiouAEIOU"
    count=0
    
    for char in name:
        if char in vowels:
            count+=1
    return count
    
vowels('hello python')
```
