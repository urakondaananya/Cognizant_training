```python
class Student:
    def __init__(self,name,age):
        self.name=name
        self.age=age

s1=Student("ananya",19)
print(s1.name)
print(s1.age)
```

```python
class Employee:
    def __init__(self,name,id,salary):
        self.name=name
        self.salary=salary
        self.id=id

s1=Employee("ananya",9,80000)
print(s1.name)
print(s1.salary)
print(s1.id)
```

```python
class Car:
    def __init__(self,brand,model):
        self.model=model
        self.brand=brand
       
s1=Car("toyota",2025)
print(s1.brand)
print(s1.model)
```

```python
class Car:
    def __init__(self,brand,model):
        self.model=model
        self.brand=brand
       
    def display_details(self):
     print('model:',self.model)
     print('brand:',self.brand)

c1=Car("toyota",2025)
c1.display_details()
```

```python
class Person:
    def func2(self):
        print("this is a person")

class Student(Person):
    def func3(self):
        print("this is a student")

object=Student()
object.func2()
object.func3()
```

```python
class Vechile:
    def func2(self):
        print("this is a vechile")

class Car(Vechile):
    def func3(self):
        print("this is a car")

object=Car()
object.func2()
object.func3()
```

```python
class GrandFather:
    def func(self):
        print("this is GrandFather")

class Father(GrandFather):
    def func(self):
        super().func()
        print("this is father")

class Child(Father):
    def func(self):
        super().func()
        print("this is child")

object=Child()
object.func()
```
