```python
students=[("anu",60),("kinnu",50),("renny",35)]
result=sorted(students,key=lambda u:u[1])
print(result)
```

```python
students=[("anu",60),("kinnu",50),("renny",35)]
result=sorted(students,key=lambda u:u[0])
print(result)
```

```python
numbers=[3,4,5,6,7]
result=list(map(lambda n:n+n,numbers))
print(result)
```

```python
numbers=[3,4,5,6,7]
result=list(map(lambda n:n*n*n,numbers))
print(result)
```

```python
numbers=[9,8,7,6,5]
result=list(filter(lambda y:y%2==0,numbers))
print(result)
```

```python
numbers=[9,7,8,4,1,7]
result=list(filter(lambda y:y%2==1,numbers))
print(result)
```

```python
from functools import reduce
num=[2,8,0,5,7,4,2]
result=reduce(lambda x,y:x+y,num)
print(result)
```
