```python
import pandas as pd
import numpy as np
s=pd.Series([10,20,30,np.nan,40,np.nan])
print(s)
print(s.isnull())# to find if any element is null or not
print(s.isnull().sum())
```

```python
import pandas as pd
import numpy as np
s=pd.Series([10,20,30,np.nan,40,np.nan])
print(s.fillna(0))
```

```python
s=pd.Series(['anu','sunny','funny','honey','kitty'])
print([s.str.upper()])
print([s.str.lower()])
```
